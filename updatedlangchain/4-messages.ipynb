```python
import os 
from dotenv import load_dotenv
load_dotenv()

from langchain.chat_models import init_chat_model
os.environ['GROQ_API_KEY']=os.getenv("GROQ_API_KEY")
model=init_chat_model("groq:qwen/qwen3-32b")
```

```python
model.invoke("Please tell what is artificial intelligence")
```

```python
from langchain.messages import SystemMessage,HumanMessage,AIMessage

messages=[
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem in artificial intelligence")
]

response=model.invoke(messages)
response.content
```

```python
system_msg=SystemMessage("You are the helpful coding assistant.")

messages=[
    system_msg,
    HumanMessage("How do I create a Rest API?")
]
response=model.invoke(messages)
print(response.content)
```

```python
### Detailed info to the  LLM through system message
from langchain.messages import SystemMessage ,HumanMessage
system_msg=SystemMessage("""
You are a senior python developer with expertise in web frameworks.
Always provide code examples and explain you reasoning.
Be consise but although in your explanations.
""")
messages=[
    system_msg,
    HumanMessage("How do I create a REST API?")
]
response=model.invoke(messages)
print(response.content)
```

```python
## message Metadata
human_msg=HumanMessage(
    content="Hello",
    name="Alice",
    id="msg_123"
)
```

```python
response=model.invoke([human_msg])
print(human_msg.content)
```

```python
from langchain.messages import AIMessage,SystemMessage,HumanMessage
#create an AI message manually (e.g for conversion history)
ai_msg=AIMessage("I'd be happy to help you with that questions!")
#add to conversion history
messages=[
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great! what's 2+2?")
]
response=model.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

```python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

#After a models make a tool call
#(Here, we demonstrate manually creating the message foor the brevity)
ai_message=AIMessage(
    content=[],
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"San Francisco"},
        "id":"call_123"

    }]
)

#Execute tool and create result message 
weather_result="Sunny,72 degree faranheit"
tool_message=ToolMessage(
    content=weather_result,
    tool_call_id='call_123'
)


#continue conversation
messages=[
    HumanMessage("What's the weather in San Francisco"),
    ai_message,
    tool_message
]
response=model.invoke(messages)
```
