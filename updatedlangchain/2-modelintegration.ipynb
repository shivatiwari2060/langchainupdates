### Models integration with Google gemini and GROQ

```python
import os 
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model=init_chat_model(model_name="gpt-4.1")
model
```

```python
## invoke the model
response=model.invoke("Hello How are you?")
response
```

```python
###chatOpenAI
from langchain_openai import ChatOpenAI
ChatOpenAI(model="gpt-4.1")
response=model.invoke("Hello ,How are you?")
response
```

### Google gemini model integration


```python
import os
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
model=init_chat_model("google_genai:gemini-2.0-flash")
response=model.invoke("Why do parrots talks?")
response
```

### Groq Model Integration


```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
model=init_chat_model("groq:qwen/qwen3-32b")
response=model.invoke("Why do parrot talk?")
response
```

```python
import sys
print(sys.executable)
```

```python
for chunk in model.stream("Why do parrots talk?"):
    print(chunk.content, end="", flush=True)
```

```python
prompts = [
    "Why do parrots have colorful feathers?",
    "Why do airplanes fly?",
    "What is quantum computing?",
]

responses = model.batch(prompts,config={
    'max_concurrency':5      #limit to 5 parallel calls
})
print(responses)

# for prompt, r in zip(prompts, responses):
#     print(f"Q: {prompt}")
#     print(f"A: {r.content}\n")
```
